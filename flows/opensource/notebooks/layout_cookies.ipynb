## 2 - Vista completa

Los dos bloques dentro de sus **guard rings**: arriba el espejo PMOS (en su
nwell), abajo el par diferencial NMOS + el bias. Cada transistor es un
**arreglo de dedos** (el `m=4` del netlist).

```python
dibuja({n: True for n in ORDEN}, 'completo')
```

## 3 - Enciende/apaga capas y haz zoom

- Apaga **met1/met2/met3** para ver solo los transistores (diff + poly).
- Deja solo **poly** y **diff**: donde el **poly (rojo) cruza la diff (verde)**
  nace un transistor. Cuenta los dedos: veras grupos de 4 (el `m=4`).
- Zoom a **bloque NMOS**: el par diferencial M3/M4 **interdigitado** + dummies.

```python
def explora(nwell=True, diff=True, tap=True, li1=True, poly=True,
            met1=True, met2=True, met3=True, zoom='completo'):
    dibuja(dict(nwell=nwell, diff=diff, tap=tap, li1=li1, poly=poly,
                met1=met1, met2=met2, met3=met3), zoom)

interact(explora,
         zoom=Dropdown(options=list(ZOOM), value='completo', description='zoom'));
```

## 4 - Ver el matching

Haz **zoom = bloque NMOS** y deja solo **diff + poly**. Lo que buscas:

1. **Dedos iguales**: M3 y M4 son el mismo dibujo repetido (misma W/L por dedo).
2. **Interdigitacion**: los dedos de M3 y M4 van **intercalados**, no en bloques
   separados -- asi ambos 've' el mismo entorno y el gradiente de proceso se
   promedia.
3. **Dummies** en los bordes: dedos que no conducen, para que los dedos reales
   de las orillas tengan vecinos identicos.

Esto es lo que en 🧅 Bunuelo era solo `m=4` en una linea de texto. Aqui es
**fisico**: por eso el layout analogico se dibuja a mano.
